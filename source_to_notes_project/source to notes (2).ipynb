```python

#created a page lookup
page_text_lookup = {
    row["page_number"]: row["page_text"]
    for row in pages
}

print("Page lookup created!")
```

```python
#create the checker 
def validate_note(row):

    page_number = row["source_page_number"]
    evidence = str(row["exact_evidence_snippet"]).strip()

    if pd.isna(page_number):
        return "Needs Review", "Missing page number."

    try:
        page_number = int(page_number)
    except:
        return "Needs Review", "Invalid page number."

    if page_number not in page_text_lookup:
        return "Needs Review", "Page does not exist."

    page_text = page_text_lookup[page_number]

    if evidence not in page_text:
        return "Needs Review", "Evidence not found on page."

    return "Supported", "Evidence verified."
```

```python
#Define the checker
def validate_note(row):
    page_number = row["source_page_number"]
    evidence = str(row["exact_evidence_snippet"]).strip()

    if pd.isna(page_number):
        return "Needs Review", "Missing page number."

    try:
        page_number = int(page_number)
    except (TypeError, ValueError):
        return "Needs Review", "Invalid page number."

    if page_number not in page_text_lookup:
        return "Needs Review", "Page does not exist."

    page_text = page_text_lookup[page_number]

    if evidence not in page_text:
        return "Needs Review", "Evidence not found on page."

    return "Supported", "Evidence verified."
```

```python
#apply the checker 
results = notes_df.apply(validate_note, axis=1)

notes_df["status"] = [r[0] for r in results]

notes_df["automatic_check_comment"] = [r[1] for r in results]

notes_df
```
